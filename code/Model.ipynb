```python
!pip install pmdarima arch xgboost scikit-learn tensorflow -q

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')

from google.colab import drive
drive.mount('/content/drive')

print("Libraries loaded successfully.")

# STEP 1: LOAD FEATURE-ENGINEERED DATASET

df = pd.read_csv(
    '/content/drive/MyDrive/MSc_Project/data/processed/usd_lkr_gold_features.csv',
    index_col='Date',
    parse_dates=True
)

print(f"Dataset loaded: {df.shape[0]} rows | {df.shape[1]} columns")
print(f"Date range: {df.index.min().date()} to {df.index.max().date()}")

# STEP 2: TRAIN / VALIDATION / TEST SPLIT

TRAIN_END = '2018-12-31'
VAL_END   = '2021-12-31'

train = df[df.index <= TRAIN_END]
val   = df[(df.index > TRAIN_END) & (df.index <= VAL_END)]
test  = df[df.index > VAL_END]

print(f"\nTRAIN: {train.index.min().date()} to {train.index.max().date()} | {len(train)} rows")
print(f"VAL:   {val.index.min().date()} to {val.index.max().date()} | {len(val)} rows")
print(f"TEST:  {test.index.min().date()} to {test.index.max().date()} | {len(test)} rows")

# STEP 3: SCALING

from sklearn.preprocessing import MinMaxScaler

TARGET_RETURN = 'target_lkr_returns'
TARGET_PRICE  = 'target_lkr_close'
FEATURE_COLS  = [c for c in df.columns if c not in [TARGET_RETURN, TARGET_PRICE]]

# Scaler fitted on training set ONLY - applied to val and test without re-fitting
scaler_X       = MinMaxScaler()
scaler_y_ret   = MinMaxScaler()
scaler_y_price = MinMaxScaler()

X_train = scaler_X.fit_transform(train[FEATURE_COLS])
X_val   = scaler_X.transform(val[FEATURE_COLS])
X_test  = scaler_X.transform(test[FEATURE_COLS])

y_train_ret   = scaler_y_ret.fit_transform(train[[TARGET_RETURN]])
y_val_ret     = scaler_y_ret.transform(val[[TARGET_RETURN]])
y_test_ret    = scaler_y_ret.transform(test[[TARGET_RETURN]])

y_train_price = scaler_y_price.fit_transform(train[[TARGET_PRICE]])
y_val_price   = scaler_y_price.transform(val[[TARGET_PRICE]])
y_test_price  = scaler_y_price.transform(test[[TARGET_PRICE]])

print(f"\nScaling complete.")
print(f"X_train: {X_train.shape} | X_val: {X_val.shape} | X_test: {X_test.shape}")

# STEP 4: EVALUATION METRICS FUNCTION

def evaluate(actual, predicted, model_name):
    actual    = np.array(actual).flatten()
    predicted = np.array(predicted).flatten()
    rmse = np.sqrt(np.mean((actual - predicted) ** 2))
    mae  = np.mean(np.abs(actual - predicted))
    mask = actual != 0 #a safety check, it skips any rows where the actual value is zero to avoid dividing by zero
    mape = np.mean(np.abs((actual[mask] - predicted[mask]) / actual[mask])) * 100
    print(f"\n{'─'*55}")
    print(f"  {model_name}")
    print(f"{'─'*55}")
    print(f"  RMSE : {rmse:.6f}") #Root Mean Squared Error, rmse = np.sqrt(np.mean((actual - predicted) ** 2))
    print(f"  MAE  : {mae:.6f}") #Mean Absolute Error, mae = np.mean(np.abs(actual - predicted))
    print(f"  MAPE : {mape:.4f}%") #Mean Absolute Perecentage Error
    print(f"{'─'*55}")
    return {'model': model_name, 'rmse': round(rmse,6),
            'mae': round(mae,6), 'mape': round(mape,4)}

results = [] # to create a full table with all model results

# STEP 5: ARIMA — UNIVARIATE BASELINE (FAST BATCH FORECAST)

from pmdarima import auto_arima

print("\n" + "="*60)
print("ARIMA — Univariate Baseline")
print("="*60)

lkr_ret_train = train['lkr_returns']
lkr_ret_val   = val['lkr_returns']
lkr_ret_test  = test['lkr_returns']

# Fit on training set
print("Searching for optimal ARIMA order:")
arima_model = auto_arima(
    lkr_ret_train,
    d=1,
    start_p=0, max_p=5,
    start_q=0, max_q=5,
    information_criterion='aic',
    stepwise=True,
    seasonal=False,
    error_action='ignore',
    suppress_warnings=True
)
print(f"Best ARIMA order: {arima_model.order}")
print(arima_model.summary())

# FAST BATCH FORECAST
# Predict all 1,106 test days in one call
# Update model with validation data first so it has full history
arima_model.update(list(lkr_ret_val))
arima_preds  = arima_model.predict(n_periods=len(lkr_ret_test))
arima_actual = lkr_ret_test.values

res_arima = evaluate(arima_actual, arima_preds, 'ARIMA')
results.append(res_arima)

# Plot
plt.figure(figsize=(14, 4))
plt.plot(test.index, arima_actual, label='Actual',
         color='#1f77b4', linewidth=0.8)
plt.plot(test.index, arima_preds,  label='ARIMA Predicted',
         color='orange', linewidth=0.8, alpha=0.8)
plt.title('ARIMA — Predicted vs Actual LKR Log Returns (Test Set 2022–2026)')
plt.xlabel('Date')
plt.ylabel('Log Return')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('/content/drive/MyDrive/MSc_Project/outputs/06a_arima_predictions.png',
            dpi=150, bbox_inches='tight')
plt.show()
print("ARIMA done.")

# STEP 6: ARIMAX - WITH GOLD AS EXOGENOUS VARIABLE (FAST BATCH FORECAST)

from pmdarima import ARIMA as pmdARIMA

print("\n" + "="*60)
print("ARIMAX — With Gold Exogenous Variable")
print("="*60)

EXOG_COLS  = ['gold_returns', 'gold_returns_lag3', 'gold_returns_lag5']
exog_train = train[EXOG_COLS]
exog_val   = val[EXOG_COLS]
exog_test  = test[EXOG_COLS]

# Fit ARIMAX with same order as ARIMA
arimax_model = pmdARIMA(order=arima_model.order)
arimax_model.fit(lkr_ret_train, X=exog_train.values)
print(arimax_model.summary())

# FAST BATCH FORECAST
arimax_model.update(list(lkr_ret_val), X=exog_val.values)
arimax_preds = arimax_model.predict(
    n_periods=len(lkr_ret_test),
    X=exog_test.values
)

res_arimax = evaluate(arima_actual, arimax_preds, 'ARIMAX')
results.append(res_arimax)

# Plot ARIMA vs ARIMAX
plt.figure(figsize=(14, 4))
plt.plot(test.index, arima_actual,  label='Actual',
         color='#1f77b4', linewidth=0.8)
plt.plot(test.index, arima_preds,   label='ARIMA',
         color='orange', linewidth=0.8, alpha=0.7)
plt.plot(test.index, arimax_preds,  label='ARIMAX',
         color='green',  linewidth=0.8, alpha=0.8)
plt.title('ARIMA vs ARIMAX - LKR Log Returns (Test Set 2022–2026)')
plt.xlabel('Date')
plt.ylabel('Log Return')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('/content/drive/MyDrive/MSc_Project/outputs/06b_arimax_predictions.png',
            dpi=150, bbox_inches='tight')
plt.show()
print("ARIMAX done.")

# STEP 7: GARCH(1,1) — VOLATILITY MODELLING

from arch import arch_model

print("\n" + "="*60)
print("GARCH(1,1) - Volatility Modelling")
print("="*60)

# Fit GARCH on ARIMA residuals (×100 for numerical stability)
# residuals - gap between what it predicted and what actually happened(the portion of daily movement that ARIMA could not explain)
arima_residuals = arima_model.resid() * 100 #multy by 100 for numeircal stability

garch_model = arch_model(arima_residuals, vol='Garch', p=1, q=1, dist='normal')
garch_fit   = garch_model.fit(disp='off')
print(garch_fit.summary())

# Batch volatility forecast for test period
garch_forecast = garch_fit.forecast(horizon=len(lkr_ret_test), reindex=False)
garch_vol      = np.sqrt(garch_forecast.variance.values[-1]) / 100

# GARCH does not make return predictions — store volatility for ensemble use
# Save as a column for Chapter 7 hybrid ensemble
garch_vol_series = pd.Series(garch_vol, index=test.index, name='garch_vol')
garch_vol_series.to_csv(
    '/content/drive/MyDrive/MSc_Project/outputs/garch_volatility.csv'
)

# Plot conditional volatility
plt.figure(figsize=(14, 4))
plt.plot(test.index, garch_vol, color='red', linewidth=0.8)
plt.title('GARCH(1,1) - Conditional Volatility Forecast (Test Set 2022–2026)')
plt.xlabel('Date')
plt.ylabel('Conditional Std Dev')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('/content/drive/MyDrive/MSc_Project/outputs/06c_garch_volatility.png',
            dpi=150, bbox_inches='tight')
plt.show()

# Use ARIMA predictions as the point forecast for GARCH evaluation
res_garch = evaluate(arima_actual, arima_preds, 'GARCH(1,1)')
results.append(res_garch)
print("GARCH done.")

# STEP 8: SEQUENCE BUILDER FOR LSTM AND GRU

def build_sequences(X, y, lookback=30):
    Xs, ys = [], []
    for i in range(len(X) - lookback):
        Xs.append(X[i:i+lookback])
        ys.append(y[i+lookback])
    return np.array(Xs), np.array(ys)

# Why 30 - It matches with the rolling window features(SMA-30,lkr_vol_30)
# computed infeature engineering,the model sees the same 30-day horizon
# that those features were calculated over.
LOOKBACK = 30

X_tr_seq, y_tr_seq = build_sequences(X_train, y_train_ret.flatten(), LOOKBACK)
X_vl_seq, y_vl_seq = build_sequences(X_val,   y_val_ret.flatten(),   LOOKBACK)
X_te_seq, y_te_seq = build_sequences(X_test,  y_test_ret.flatten(),  LOOKBACK)

#shape(samples, consecutive days, features)
print(f"\nSequence shapes:")
print(f"  X_train: {X_tr_seq.shape} | y_train: {y_tr_seq.shape}")
print(f"  X_val:   {X_vl_seq.shape} | y_val:   {y_vl_seq.shape}")
print(f"  X_test:  {X_te_seq.shape} | y_test:  {y_te_seq.shape}")

# Actual returns for evaluation (inverse transform)
actual_returns_test = scaler_y_ret.inverse_transform(
    y_te_seq.reshape(-1, 1)
).flatten()

test_dates_seq = test.index[LOOKBACK:]

# STEP 9: LSTM MODEL - nenural network with built in memory

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, GRU, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

print("\n" + "="*60)
print("LSTM - Long Short-Term Memory")
print("="*60)

tf.random.set_seed(42)
np.random.seed(42)

n_features = X_tr_seq.shape[2]

lstm_model = Sequential([
    # Layer 1: 64 memory cells, passes output forward
    LSTM(64, input_shape=(LOOKBACK, n_features), return_sequences=True),
    # Randomly switches off 20% of neurons - prevents overfitting
    Dropout(0.2),
    # Layer 2: 32 memory cells, produces final output
    LSTM(32, return_sequences=False),
    # Another 20% dropout
    Dropout(0.2),
    # Fully connected layer — learns non-linear combinations
    Dense(16, activation='relu'),
    # Output: one number — tomorrow's LKR log return
    Dense(1)
], name='LSTM_Model')

lstm_model.compile(
    optimizer=tf.keras.optimizers.RMSprop(learning_rate=0.001),
    loss='mse'
)
lstm_model.summary()

callbacks = [

    # Why patience=10-If the validation loss stops improving for 10 consecutive epochs,
    # training stops automatically and the best weights are restored.
    # This prevents the model from memorising the training data
    # rather than learning genuine patterns.

    EarlyStopping(monitor='val_loss', patience=10, restore_best_weights=True),

    # If progress stalls for 5 epochs, the learning rate is cut in
    # half — helping the model make finer adjustments as it gets
    # closer to the optimal solution

    ReduceLROnPlateau(monitor='val_loss', factor=0.5, patience=5, min_lr=1e-6)
]

lstm_history = lstm_model.fit(
    X_tr_seq, y_tr_seq,
    validation_data=(X_vl_seq, y_vl_seq),
    epochs=100,
    batch_size=32,
    callbacks=callbacks,
    verbose=1
)

# Predict and inverse transform
# Come out in the scale[0,1] range using inverse_transform converts
# them back to real log return values for eavluation
lstm_preds = scaler_y_ret.inverse_transform(
    lstm_model.predict(X_te_seq)
).flatten()

res_lstm = evaluate(actual_returns_test, lstm_preds, 'LSTM')
results.append(res_lstm)

# Training loss plot
plt.figure(figsize=(10, 4))
plt.plot(lstm_history.history['loss'],     label='Train loss')
plt.plot(lstm_history.history['val_loss'], label='Val loss')
plt.title('LSTM - Training and Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('MSE Loss')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('/content/drive/MyDrive/MSc_Project/outputs/06d_lstm_loss.png',
            dpi=150, bbox_inches='tight')
plt.show()

# Prediction plot
plt.figure(figsize=(14, 4))
plt.plot(test_dates_seq, actual_returns_test,
         label='Actual', color='#1f77b4', linewidth=0.8)
plt.plot(test_dates_seq, lstm_preds,
         label='LSTM Predicted', color='#ff7f0e', linewidth=0.8, alpha=0.8)
plt.title('LSTM - Predicted vs Actual LKR Log Returns (Test Set 2022–2026)')
plt.xlabel('Date')
plt.ylabel('Log Return')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('/content/drive/MyDrive/MSc_Project/outputs/06e_lstm_predictions.png',
            dpi=150, bbox_inches='tight')
plt.show()

lstm_model.save('/content/drive/MyDrive/MSc_Project/models/lstm_model.h5')
print("LSTM done.")

# STEP 10: GRU MODEL - lighter version of LSTM(only two gates-reset, update)

print("\n" + "="*60)
print("GRU - Gated Recurrent Unit")
print("="*60)

tf.random.set_seed(42)
np.random.seed(42)

#same as LSTM , only difference is cell type
gru_model = Sequential([
    GRU(64, input_shape=(LOOKBACK, n_features), return_sequences=True),
    Dropout(0.2),
    GRU(32, return_sequences=False),
    Dropout(0.2),
    Dense(16, activation='relu'),
    Dense(1)
], name='GRU_Model')

gru_model.compile(
    optimizer=tf.keras.optimizers.RMSprop(learning_rate=0.001),
    loss='mse'
)
gru_model.summary()

gru_history = gru_model.fit(
    X_tr_seq, y_tr_seq,
    validation_data=(X_vl_seq, y_vl_seq),
    epochs=100,
    batch_size=32,
    callbacks=callbacks,
    verbose=1
)

# Predict and inverse transform
gru_preds = scaler_y_ret.inverse_transform(
    gru_model.predict(X_te_seq)
).flatten()

res_gru = evaluate(actual_returns_test, gru_preds, 'GRU')
results.append(res_gru)

# Training loss plot
plt.figure(figsize=(10, 4))
plt.plot(gru_history.history['loss'],     label='Train loss')
plt.plot(gru_history.history['val_loss'], label='Val loss')
plt.title('GRU - Training and Validation Loss')
plt.xlabel('Epoch')
plt.ylabel('MSE Loss')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('/content/drive/MyDrive/MSc_Project/outputs/06f_gru_loss.png',
            dpi=150, bbox_inches='tight')
plt.show()

# Prediction plot
plt.figure(figsize=(14, 4))
plt.plot(test_dates_seq, actual_returns_test,
         label='Actual', color='#1f77b4', linewidth=0.8)
plt.plot(test_dates_seq, gru_preds,
         label='GRU Predicted', color='green', linewidth=0.8, alpha=0.8)
plt.title('GRU - Predicted vs Actual LKR Log Returns (Test Set 2022–2026)')
plt.xlabel('Date')
plt.ylabel('Log Return')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('/content/drive/MyDrive/MSc_Project/outputs/06g_gru_predictions.png',
            dpi=150, bbox_inches='tight')
plt.show()

gru_model.save('/content/drive/MyDrive/MSc_Project/models/gru_model.h5')
print("GRU done.")

# STEP 11: XGBOOST
# Builds a sequence of decision trees where each tree learns from
# the mistakes of the previous one,looks at one row at a time.

import xgboost as xgb

print("\n" + "="*60)
print("XGBoost - Gradient Boosted Trees")
print("="*60)

# Flattens the target arrays from 2D into 1D-XGBoost expects
# a flat array,not a matrix
# Uses the full X_train, X_val, X_test directly-no sequence
# reshaping neede unlike LSTM and GRU

y_tr_xgb = y_train_ret.flatten()
y_vl_xgb = y_val_ret.flatten()
y_te_xgb = y_test_ret.flatten()

xgb_model = xgb.XGBRegressor(
    #maximum number of trees to build
    n_estimators=500,
    #how many splits each tree can make(reduce overfitting)
    max_depth=4,
    #how much each tree corrects the previous-small value=careful learning
    learning_rate=0.05,
    #each tree only sees 80% of rows(adds randomness, reduces overfiting)
    subsample=0.8,
    #each tree only sees 80% features(same reason as above)
    colsample_bytree=0.8,
    #minimum obersvations per leaf(prevents overfitting on tiny groups)
    min_child_weight=5,
    #pushes less useful feature weights toward zero
    reg_alpha=0.1,
    #penalises large feature weights, keeps model conservative
    reg_lambda=1.0,
    #stops if validation RMSE does not improve for 50 consecutive trees
    early_stopping_rounds=50,
    eval_metric='rmse',
    random_state=42,
    verbosity=1
)

xgb_model.fit(
    X_train, y_tr_xgb,
    eval_set=[(X_val, y_vl_xgb)],
    # verbose=50, prints progress every 50trees then can see whether the model is still improving
    verbose=50
)

print(f"\nBest iteration: {xgb_model.best_iteration}")

# Predict and inverse transform
xgb_preds = scaler_y_ret.inverse_transform(
    xgb_model.predict(X_test).reshape(-1, 1)
).flatten()

actual_ret_xgb = scaler_y_ret.inverse_transform(
    y_te_xgb.reshape(-1, 1)
).flatten()

#call evaluate to compare RMSE,MAE and MAPE and add to results table
res_xgb = evaluate(actual_ret_xgb, xgb_preds, 'XGBoost')
results.append(res_xgb)

# Feature importance - top 20
plt.figure(figsize=(12, 8))
feat_imp = pd.Series(xgb_model.feature_importances_, index=FEATURE_COLS)
feat_imp.nlargest(20).sort_values().plot(kind='barh', color='#1f77b4')
plt.title('XGBoost — Top 20 Feature Importances')
plt.xlabel('Importance Score')
plt.tight_layout()
plt.savefig('/content/drive/MyDrive/MSc_Project/outputs/06h_xgboost_feature_importance.png',
            dpi=150, bbox_inches='tight')
plt.show()

# Prediction plot
plt.figure(figsize=(14, 4))
plt.plot(test.index, actual_ret_xgb,
         label='Actual', color='#1f77b4', linewidth=0.8)
plt.plot(test.index, xgb_preds,
         label='XGBoost Predicted', color='purple', linewidth=0.8, alpha=0.8)
plt.title('XGBoost — Predicted vs Actual LKR Log Returns (Test Set 2022–2026)')
plt.xlabel('Date')
plt.ylabel('Log Return')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.savefig('/content/drive/MyDrive/MSc_Project/outputs/06i_xgboost_predictions.png',
            dpi=150, bbox_inches='tight')
plt.show()
print("XGBoost done.")

# STEP 12: SAVE ALL PREDICTIONS FOR CHAPTER 7 HYBRID ENSEMBLE

# Align all predictions to the same index length (test set length)
# ARIMA/ARIMAX use full test index; LSTM/GRU use test_dates_seq (30 rows shorter)
# Align on the shorter window for fair ensemble comparison

n_ensemble = len(test_dates_seq)

preds_df = pd.DataFrame({
    'actual':         actual_returns_test,
    'arima':          arima_preds[-n_ensemble:],
    'arimax':         arimax_preds[-n_ensemble:],
    'lstm':           lstm_preds,
    'gru':            gru_preds,
    'xgboost':        xgb_preds[-n_ensemble:],
    'garch_vol':      garch_vol[-n_ensemble:],
}, index=test_dates_seq)

preds_df.to_csv(
    '/content/drive/MyDrive/MSc_Project/outputs/06_all_predictions.csv'
)
print("\nAll predictions saved to: outputs/06_all_predictions.csv")
print("This file will be loaded in Chapter 7 for the hybrid ensemble.")

# STEP 13: INDIVIDUAL MODEL RESULTS SUMMARY

print("\n" + "="*60)
print("INDIVIDUAL MODEL RESULTS — TEST SET (2022–2026)")
print("="*60)

results_df = pd.DataFrame(results)
results_df = results_df.sort_values('rmse').reset_index(drop=True)
print(results_df.to_string(index=False))

results_df.to_csv(
    '/content/drive/MyDrive/MSc_Project/outputs/06_model_results.csv',
    index=False
)
print("\nResults saved to: outputs/06_model_results.csv")

# Bar chart comparison
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
fig.suptitle('Individual Model Comparison — Test Set (2022–2026)',
             fontsize=13, fontweight='bold')

colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728', '#9467bd', '#8c564b']

for ax, metric, label in zip(
    axes,
    ['rmse', 'mae', 'mape'],
    ['RMSE', 'MAE', 'MAPE (%)']
):
    bars = ax.bar(
        results_df['model'],
        results_df[metric],
        color=colors[:len(results_df)]
    )
    ax.set_title(label)
    ax.set_ylabel(label)
    ax.tick_params(axis='x', rotation=45)
    ax.grid(True, alpha=0.3, axis='y')
    for bar, val in zip(bars, results_df[metric]):
        ax.text(
            bar.get_x() + bar.get_width() / 2,
            bar.get_height(),
            f'{val:.4f}',
            ha='center', va='bottom', fontsize=8
        )

plt.tight_layout()
plt.savefig('/content/drive/MyDrive/MSc_Project/outputs/06j_model_comparison.png',
            dpi=150, bbox_inches='tight')
plt.show()

print("ALL MODELS COMPLETE")
```
